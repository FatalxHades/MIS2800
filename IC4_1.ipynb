```python
%%sql
/* 1. Create the table with STRICT types */
CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    price REAL NOT NULL,
    category TEXT
) STRICT;

/* 2. Insert at least 3 items (Create operation) */
INSERT INTO products (name, price, category)
VALUES ('Superbowl Jersey', 120.00, 'Apparel');

INSERT INTO products (name, price, category)
VALUES ('Wireless Headphones', 89.99, 'Electronics');

INSERT INTO products (name, price, category)
VALUES ('Water Bottle', 24.50, 'Accessories');
```

```python
%%sql
/* Show all products that cost more than $50 */
SELECT *
FROM products
WHERE price > 50;
```

```python
%%sql
/* Update a price */
UPDATE products
SET price = 99.99
WHERE name = 'Superbowl Jersey';

/* Delete an item */
DELETE FROM products
WHERE name = 'Water Bottle';
```
